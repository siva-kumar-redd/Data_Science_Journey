```python
import numpy as np

values = np.array([2,4,6])

population_variance = np.var(values)
print(population_variance)
```

Sample variance using pandas

```python
import pandas as pd
values = pd.Series([2,4,6])
print(values.var())
```

Numpy: Population variance Vs sample variance

```python
values = np.array([2,4,6])
poplulation_var = np.var(values)
sample_var = np.var(values,ddof=1)
print(poplulation_var)
print(sample_var)
```

Pandas: Population variance Vs sample variance

```python
values = pd.Series([2,4,6])
poplulation_var = values.var()
sample_var = values.var(ddof=0)
print(poplulation_var)
print(sample_var)
```

Sales Variability

```python
sales = pd.Series([
    10000, 10200, 9800, 10100, 9900,
    10050, 10300
])

print("Mean Sales:", sales.mean())
print("Sample Variance:", sales.var())
print("Population Variance:", sales.var(ddof=0))
```

Compare Two Datasets

```python
stable = pd.Series([48, 49, 50, 51, 52])
variable = pd.Series([10, 30, 50, 70, 90])

print("Stable Variance:", stable.var(ddof=0))
print("Variable Variance:", variable.var(ddof=0))
```

Mini Data Science Challenge

```python
data = pd.DataFrame({
    "Department": ["A", "A", "A", "B", "B", "B"],
    "Salary": [30000, 31000, 30500, 25000, 50000, 80000]
})

print(data.groupby("Department")["Salary"].var())
```
