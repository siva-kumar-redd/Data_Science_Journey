Probability Distribution with Pandas

```python
import pandas as pd

data = pd.DataFrame({
    "Purchases": [0, 1, 2, 3],
    "Probability": [0.40, 0.35, 0.20, 0.05]
})

print(data)

expected_value = data["Purchases"] * data["Probability"]
print(expected_value)
```

Verify the Probability Distribution

```python
total_probability = data["Probability"].sum()

print(total_probability)
```

Expected Value

```python
expected_value = (
    data["Purchases"] * data["Probability"]
).sum()

print("Expected Value:", expected_value)
```

Variance of the Random Variable

```python
data["Squared_Deviation"] = (
    data["Purchases"] - expected_value
) ** 2

data["Weighted_Deviation"] = (
    data["Squared_Deviation"] * data["Probability"]
)

variance = data["Weighted_Deviation"].sum()

print("Variance:", variance)
```

Standard Deviation

```python
import numpy as np

standard_deviation = np.sqrt(variance)

print("Standard Deviation:", standard_deviation)
```
