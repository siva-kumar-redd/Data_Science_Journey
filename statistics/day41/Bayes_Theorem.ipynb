Bayes Theorem with NumPy

```python
import numpy as np
import pandas as pd
```

```python
p_a = 0.20
p_b_given_a = 0.50
p_b = 0.25

p_a_given_b = (p_b_given_a * p_a) / p_b

print("P(A|B):", p_a_given_b)
```

Fraud Detection with NumPy

```python
p_fraud = 0.05
p_alert_given_fraud = 0.80
p_alert = 0.10

p_fraud_given_alert = (
    p_alert_given_fraud * p_fraud
) / p_alert

print("P(Fraud|Alert):", p_fraud_given_alert)
```

Bayes with Pandas

```python
import pandas as pd

data = pd.DataFrame({
    "Fraud": [
        True, True, True, False, False,
        False, False, False, False, False
    ],
    "Alert": [
        True, True, False, True, False,
        False, True, False, False, False
    ]
})

print(data)
```

```python
p_fraud = data["Fraud"].mean()
print("P(Fraud):", p_fraud)
```

```python
fraud_tranctions = data[data["Fraud"] == True]
p_alert_given_fraud = fraud_tranctions["Alert"].mean()
print("P(Alert|Fraud):", p_alert_given_fraud)
```

```python
p_alert = data["Alert"].mean()
print("P(Alert):", p_alert)
```

```python
p_fraud_given_alert = (
    p_alert_given_fraud * p_fraud
) / p_alert

print("P(Fraud|Alert):", p_fraud_given_alert)
```
