Probability Analysis

```python
import numpy as np
import pandas as pd
```

```python
data = pd.DataFrame({
    "Customer": ["A","B","C","D","E","F","G","H","I","J"],
    "Mobile_User": [True, True, True, True, True,
                    False, False, True, False, False],
    "Purchased": [True, False, True, True, False,
                  False, True, True, False, False]
})

print(data)
```

Overall Purchase Probability

```python
purchased = data["Purchased"].sum()/data["Customer"].count()
print(purchased)
```

Mobile User Probability

```python
mobile_users = data["Mobile_User"].sum()/data["Customer"].count()
print(mobile_users)
```

Conditional Probability Using Pandas

```python
mobile_users = data[data["Mobile_User"] == True]

mobile_purchase_probability = (
    mobile_users["Purchased"].sum() / len(mobile_users)
)

print(mobile_purchase_probability)
```

NumPy Boolean Filtering

```python
import numpy as np
mobile = np.array(data["Mobile_User"])
purchased = np.array(data["Purchased"])

both = mobile & purchased

probability = both.sum()/mobile.sum()
print(probability)
```

Non-Mobile Customers

```python
non_mobile_users = data[data["Mobile_User"] == False]
non_mobile_user_probabiity = (non_mobile_users["Purchased"].sum()/len(non_mobile_users))
print(non_mobile_user_probabiity)
```

GroupBy

```python
purchase_rate = data.groupby("Mobile_User")["Purchased"].mean()

print(purchase_rate)
```
