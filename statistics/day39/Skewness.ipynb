```python
import pandas as pd

data = pd.DataFrame({
    "Customer": ["A", "B", "C", "D", "E", "F", "G", "H", "I", "J"],
    "Spending": [500, 550, 600, 620, 650, 700, 750, 800, 900, 3000]
})

mean = data["Spending"].mean()
median = data["Spending"].median()
mode = data["Spending"].mode()
print(mean)
print(median)
print(mode)
```

Quartiles

```python
q1 = data["Spending"].quantile(0.25)
q2 = data["Spending"].quantile(0.50)
q3 = data["Spending"].quantile(0.75)


print(q1)
print(q2)
print(q3)
```

IQR

```python
iqr = q3-q1
print(iqr)
```

Outlier Detection

```python
lower_bound = q1-(1.5*iqr)
upper_bound = q3+(1.5*iqr)

print(lower_bound)
print(upper_bound)

outliers = data[
    (data["Spending"]<lower_bound) | (data["Spending"]>upper_bound)
]
print(outliers)
```

Range

```python
range_data = (data["Spending"].max() - data["Spending"].min())
print(range_data)
```

Variance & Standard Deviation

```python
variance = data["Spending"].var(ddof=0)
std = data["Spending"].std(ddof=0)
print(variance)
print(std)
```

```python
data["Z_Score"] = (
    data["Spending"] - mean
) / std

print(data)
```
