Standard Deviation with NumPy

```python
import numpy as np
import pandas as pd
```

```python
values = np.array([10,20,30,40,50])
population_sd = np.std(values)
sample_sd = np.std(values,ddof=1)
print(population_sd)
print(sample_sd)
```

Standard Deviation with Pandas

```python
values = pd.Series([10, 20, 30, 40, 50])
population_sd = values.std(ddof=0)
sample_std = values.std()
print(population_sd)
print(sample_std)
```

Z-Scores

```python
values = np.array([10, 20, 30, 40, 50])
average = np.mean(values)
sd = np.std(values)
z_score = (values-average)/sd
print(average)
print(sd)
print(z_score)
```

Standardization with Pandas

```python
values = pd.Series([10, 20, 30, 40, 50])
mean = values.mean()
std = values.std(ddof=0)
standarized_data = (values-mean)/std
print(values)
print(standarized_data)
```

Mini Data Science Challenge

```python
data = pd.DataFrame({
    "Student": ["A", "B", "C", "D", "E"],
    "Math": [60, 70, 80, 90, 100],
    "Statistics": [50, 65, 75, 85, 95]
})
avg_math = data["Math"].mean()
avg_stats = data["Statistics"].mean()
math_sd = data["Math"].std(ddof=0)
stats_sd = data["Statistics"].std(ddof=0)
data["Math_Z"] = (
    data["Math"] - data["Math"].mean()
) / data["Math"].std(ddof=0)
data["Stats_Z"] = (
    data["Statistics"] - data["Statistics"].mean()
) / data["Statistics"].std(ddof=0)
print(data)
print(avg_math)
print(avg_stats)
print(math_sd)
print(data["Math_Z"])
print(data["Stats_Z"])
```
