Bernoulli Simulation

```python
import numpy as np

np.random.seed(42)

p = 0.30

purchases = np.random.binomial(
    n=1,
    p=p,
    size=1000
)

print("Total customers:", len(purchases))
print("Purchases:", purchases.sum())
print("Purchase rate:", purchases.mean())
```

Binomial Simulation

```python
daily_purchases = np.random.binomial(
    n=100,
    p=0.30,
    size=1000
)

print("Average purchases:", daily_purchases.mean())
print("Minimum purchases:", daily_purchases.min())
print("Maximum purchases:", daily_purchases.max())
```

Poisson Simulation

```python
tickets = np.random.poisson(
    lam=5,
    size=1000
)

print("Average tickets:", tickets.mean())
print("Minimum tickets:", tickets.min())
print("Maximum tickets:", tickets.max())
```

Normal Distribution Simulation

```python
delivery_time = np.random.normal(
    loc=30,
    scale=5,
    size=1000
)

print("Mean:", delivery_time.mean())
print("Standard Deviation:", delivery_time.std())
print("Minimum:", delivery_time.min())
print("Maximum:", delivery_time.max())
```

Visualize the Distributions

```python
import matplotlib.pyplot as plt

plt.hist(
    delivery_time,
    bins=30
)

plt.xlabel("Delivery Time")
plt.ylabel("Frequency")
plt.title("Simulated Delivery Time")
plt.show()
```

Compare Binomial and Poisson

```python
binomial_data = np.random.binomial(
    n=100,
    p=0.05,
    size=1000
)

poisson_data = np.random.poisson(
    lam=5,
    size=1000
)

print("Binomial Mean:", binomial_data.mean())
print("Poisson Mean:", poisson_data.mean())
```
